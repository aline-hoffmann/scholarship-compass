```python
# Lendo o arquivo "actors.csv"

def ler_csv(arquivo):
    with open(arquivo, "r", encoding="utf-8") as actors:
        linhas = actors.readlines()

    dados = []
    for linha in linhas[1:]: 
        partes = []
        atual = ""
        dentro_aspas = False
        for c in linha.strip():
            if c == '"' and not dentro_aspas:
                dentro_aspas = True
                continue
            elif c == '"' and dentro_aspas:
                dentro_aspas = False
                continue
            if c == "," and not dentro_aspas:
                partes.append(atual.strip())
                atual = ""
            else:
                atual += c
        partes.append(atual.strip())

        ator = partes[0]
        total_gross = float(partes[1])
        num_movies = int(partes[2])
        avg_movie = float(partes[3])
        best_movie = partes[4]
        gross_best = float(partes[5])

        dados.append({
            "Actor": ator,
            "Total Gross": total_gross,
            "Number of Movies": num_movies,
            "Average per Movie": avg_movie,
            "#1 Movie": best_movie,
            "Gross": gross_best
        })

    return dados

dados = ler_csv("actors.csv")
print("Carregado:", len(dados), "atores")

# Verificando uma linha
print(dados[0]) 
```

### ETAPA 1.

Verificando qual o ator com maior número de filmes e qual a quantidade.

```python
def etapa1(dados):
    ator = max(dados, key=lambda x: x["Number of Movies"])
    return f"O ator com maior número de filmes é {ator['Actor']} com {ator['Number of Movies']} filmes."

print("Etapa 1:", etapa1(dados))
```

```python
# Salvando a resposta como arquivo txt

resposta1 = etapa1(dados)

with open("etapa1.txt", "w", encoding="utf-8") as f:
    f.write(resposta1)
```

### ETAPA 2.

Verificando qual a média de receita de bilheteria bruta dos principais filmes, considerando todos os atores.

```python
# Calculando e mostrando a média

def etapa2(dados):
    media = sum(a["Gross"] for a in dados) / len(dados)
    return f"A média de receita bruta dos principais filmes, considerando todos os atores, é {media:.2f}."

print("Etapa 2:", etapa2(dados))
```

```python
# Salvando a resposta como arquivo txt

resposta2 = etapa2(dados)

with open("etapa2.txt", "w", encoding="utf-8") as f:
    f.write(resposta2)
```

### ETAPA 3.

Verificando o ator/atriz com maior média de receita de bilheteria bruta por filme.

```python
# Encontrando o ator com maior média de bilheteria

def etapa3(dados):
    ator = max(dados, key=lambda x: x["Average per Movie"])
    return f"O ator com maior média de bilheteria por filme é {ator['Actor']}, com média de {ator['Average per Movie']:.2f}."

print(etapa3(dados))
```

```python
# Salvando a resposta como arquivo txt

resposta3 = etapa3(dados)

with open("etapa3.txt", "w", encoding="utf-8") as f:
    f.write(resposta3)
```
